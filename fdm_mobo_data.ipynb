```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path
import matplotlib.colors as mcolors
import sys
import matplotlib
matplotlib.rcParams["font.family"] = "STIXGeneral"
matplotlib.rcParams['mathtext.fontset'] = 'stix'
matplotlib.rcParams["xtick.direction"] = "in"
matplotlib.rcParams["ytick.direction"] = "in"
plt.rcParams['xtick.major.size'] = 2  # Length of major ticks
plt.rcParams['ytick.major.size'] = 2  # Length of major ticks
plt.rcParams['xtick.minor.size'] = 1  # Length of minor ticks
plt.rcParams['ytick.minor.size'] = 1  # Length of minor ticks
```

```python
folder = Path(r"C:\Users\zhengyang\Downloads")
file_dir = folder / "fan-flow-h2d.csv"
data = pd.read_csv(file_dir, usecols=[0,1,2,3,4,5])
```

```python
plt.figure(figsize=(3.5, 3), dpi=300)
for fan, gfan in data.groupby("fan"):
    plt.plot(gfan["flow"], gfan["Ra"], label=f"Fan {fan}", ls="", marker="o")
plt.legend(fontsize=8, frameon=False)
plt.xlabel("Flow ratio")
plt.ylabel("Surface roughness, $Ra$ ($\\mu$m)")
plt.ylim(0, 30)
```

```python
plt.figure(figsize=(3.5, 3), dpi=300)
for flow, gflow in data.groupby("flow"):
    if flow == 1.03: continue
    plt.plot(gflow["fan"], gflow["Ra"], label=f"Flow {flow}", ls="", marker="o")
plt.legend(fontsize=8, frameon=False)
plt.xlabel("Fan power")
plt.ylabel("Surface roughness, $Ra$ ($\\mu$m)")
```

```python
plt.figure(figsize=(3.5, 3), dpi=300)
for fan, gfan in data.groupby("fan"):
    plt.plot(gfan["flow"], gfan["overhang"], label=f"Fan {fan}", ls="", marker="o")
plt.legend(fontsize=8, frameon=False)
plt.xlabel("Flow ratio")
plt.ylabel("Overhang score")
```
