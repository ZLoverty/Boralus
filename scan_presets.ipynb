```python
from pathlib import Path
import glob
import json

folder = Path("~/Documents/GitHub/Polymaker-Preset/preset").expanduser()
```

```python
str(folder / "**.json")
```

```python
l = glob.glob(str(folder /"**" / "*.json"), recursive=True)
```

```python
for file in l:
    with open(file, "r") as f:
        data = json.load(f)
        ntl = data["nozzle_temperature"]
        if len(ntl) > 0:
            nt = float(ntl[0])
            if nt < 200:
                print(ntl)
                print(file + " | " + str(nt))
```
